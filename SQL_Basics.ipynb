3. Q3. Display All Employee Records Sorted by Salary (Lowest to Highest)
Hint: Use the  
ORDER BY
clause on the salary column

-  To display all employee records sorted by salary from lowest to highest, use the ORDER BY clause with the ASC keyword (ascending order).

SELECT *
FROM employees
ORDER BY salary ASC;


Output:
The employee records will be displayed starting with the employee having the lowest salary and ending with the highest salary.

4. . Show Employees Sorted by Department (A–Z) and Salary (High → Low)

-  To display employees sorted by Department (A–Z) and then by Salary (Highest to Lowest) within each department, use:

SELECT *
FROM employees
ORDER BY department ASC, salary DESC;

Explanation:

department ASC → sorts departments alphabetically (A to Z).


salary DESC → sorts salaries from highest to lowest within each department.

5. List All Employees in the IT Department, Ordered by Hire Date (Newest First)

-  To list all employees in the IT Department and display them in order of hire date (newest first), use:

SELECT *
FROM employees
WHERE department = 'IT'
ORDER BY hire_date DESC;

Explanation:

WHERE department = 'IT' → filters only IT department employees.


ORDER BY hire_date DESC → sorts employees by hire date from newest to oldest.

9. What is the Difference Between a Primary Key and a Foreign Key?

-  | Primary Key                                 | Foreign Key                                  |
| ------------------------------------------- | -------------------------------------------- |
| Uniquely identifies each record in a table. | Links one table to another table.            |
| Cannot contain NULL values.                 | Can contain NULL values (unless restricted). |
| Values must be unique.                      | Values can be repeated.                      |
| Only one primary key is allowed per table.  | A table can have multiple foreign keys.      |
| Ensures entity integrity.                   | Ensures referential integrity.               |


Example

Employees Table

| employee_id (Primary Key) | employee_name |
| ------------------------- | ------------- |
| 101                       | Amit          |
| 102                       | Riya          |


Orders Table

| order_id | employee_id (Foreign Key) |
| -------- | ------------------------- |
| 1        | 101                       |
| 2        | 102                       |


Here:

employee_id in the Employees table is the Primary Key because it uniquely identifies each employee.
employee_id in the Orders table is the Foreign Key because it refers to the primary key in the Employees table.

In short: A Primary Key uniquely identifies a record, while a Foreign Key creates a relationship between tables.

10.  What Are Constraints in SQL and Why Are They Used?

-  What Are Constraints in SQL?

Constraints are rules applied to table columns in a database to ensure the accuracy, validity, and integrity of the data.

Why Are Constraints Used?


To prevent invalid data from being entered.
To maintain data consistency and accuracy.
To enforce relationships between tables.
To ensure data integrity in the database.
Common SQL Constraints

| Constraint    | Purpose                                        |
| ------------- | ---------------------------------------------- |
| `PRIMARY KEY` | Uniquely identifies each record in a table.    |
| `FOREIGN KEY` | Establishes a relationship between two tables. |
| `NOT NULL`    | Ensures a column cannot have NULL values.      |
| `UNIQUE`      | Ensures all values in a column are different.  |
| `CHECK`       | Ensures values meet a specific condition.      |
| `DEFAULT`     | Assigns a default value if none is provided.   |


Example

CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    name VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE,
    salary DECIMAL(10,2) CHECK (salary > 0),
    department VARCHAR(20) DEFAULT 'General'
);

In this example:

PRIMARY KEY makes employee_id unique.
NOT NULL requires a name.
UNIQUE prevents duplicate emails.
CHECK ensures salary is greater than 0.
DEFAULT sets the department to "General" if no value is provided.

Conclusion: Constraints are used to enforce rules on data, helping maintain the reliability and integrity of the database.
